8. **Perform correlation analysis among numerical variables.**

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

df = pd.read_csv("Global_Youtube_Statistics.csv", encoding="latin1")

numeric_df = df.select_dtypes(include=["number"])

corr_matrix = numeric_df.corr()

print(corr_matrix)

plt.figure(figsize=(12,8))
sns.heatmap(corr_matrix, annot=True, cmap="coolwarm", fmt=".2f")
plt.title("Correlation Analysis Among Numerical Variables")
plt.show()
```

Dark red (~0.8 to 1.0) → Very strong positive correlation.

Light red (~0.3 to 0.7) → Moderate positive correlation.

White (~0.0) → No correlation.

Light blue (~-0.3 to -0.7) → Moderate negative correlation.

Dark blue (~-0.8 to -1.0) → Very strong negative correlation.

10. **Detect outliers in subscriber counts using a box plot.**

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(8,6))
sns.boxplot(x=df['subscribers'],color="green")
plt.title("Box Plot of Subscriber Counts")
plt.ylabel("Subscribers")
plt.show()
```

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

Q1 = df["subscribers"].quantile(0.25)
Q3 = df["subscribers"].quantile(0.75)
IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

df_no_outliers = df[(df["subscribers"] >= lower_bound) & (df["subscribers"] <= upper_bound)]

df_capped = df.copy()
df_capped.loc[df_capped["subscribers"] > upper_bound, "subscribers"] = upper_bound
df_capped.loc[df_capped["subscribers"] < lower_bound, "subscribers"] = lower_bound

plt.figure(figsize=(8,6))
sns.boxplot(x=df_no_outliers["subscribers"], color="lightgreen")
plt.title("Box Plot of Subscribers")
plt.show()

plt.figure(figsize=(8,6))
sns.boxplot(x=df_capped["subscribers"], color="orange")
plt.title("Box Plot of Subscribers")
plt.show()
```

12. **Compare average monthly earnings across channel categories.**

```python
import pandas as pd
import matplotlib.pyplot as plt

df["avg_monthly_earnings"] = (df["lowest_monthly_earnings"] + df["highest_monthly_earnings"]) / 2

category_earnings = df.groupby("category")["avg_monthly_earnings"].mean().reset_index()

category_earnings = category_earnings.sort_values(by="avg_monthly_earnings", ascending=False)

print(category_earnings)

plt.figure(figsize=(8,6))
plt.bar(category_earnings["category"], category_earnings["avg_monthly_earnings"], color="teal")
plt.xticks(rotation=45, ha="right")
plt.title("Average Monthly Earnings Across Channel Categories")
plt.xlabel("Channel Category")
plt.ylabel("Average Monthly Earnings")
plt.show()
```

13. **Analyze channel creation trends over the years.**

```python
import pandas as pd
import matplotlib.pyplot as plt

creation_trends = df.groupby("created_year")["Youtuber"].count().reset_index()

creation_trends.columns = ["Year", "Channels_Created"]

creation_trends = creation_trends.sort_values(by="Year")

print(creation_trends)

plt.figure(figsize=(8,6))
plt.plot(creation_trends["Year"], creation_trends["Channels_Created"], marker="o", color="darkgreen")
plt.title("YouTube Channel Creation Trends Over the Years")
plt.xlabel("Year")
plt.ylabel("Number of Channels Created")
plt.grid(True)
plt.show()
```

14. **Summarize the key findings and business insights.**

**Popularity and Growth**

**What people watch:** Music and entertainment are the absolute favorites, combined hitting over 5 billion subscribers. Everyday vloggers and gamers also have massive fanbases.

**Who is watching:** A few specific countries do most of the watching and subscribing, driving most of the platform's traffic.

**The creator boom:** The number of new channels being started has skyrocketed lately because everyone wants to be a content creator.*
Views and Money

**Posting vs. Views:** Generally, the more videos a channel posts, the more views it gets. However, some "lucky" channels post barely anything but get millions of views.

**The cash cows:** Music and entertainment channels make the highest average monthly profits. Niche topics like travel or pets make very little money.

**The ultimate rule:** Subscriptions and views go hand-in-hand. If a channel's subscriber count shoots up, its view count almost always does too. Posting tons of videos matters much less than getting people to subscribe.

**Business Insights**

**Where to Invest:** *Companies should put their money into the types of videos that are already proven to make big money.*

**Where to Grow:** *They should target newer, fast-growing audiences (like viewers in developing countries) to find new customers first.*
